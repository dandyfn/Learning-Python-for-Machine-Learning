## Bagian A — Praktik
### Soal 1: Rata-rata Nilai Ujian

Membuat algoritma untuk menghitung rata-rata dari sejumlah nilai ujian yang dimasukkan oleh pengguna.

### Pseudocode

INPUT n
INPUT nilai sebanyak n kali

total ← 0

FOR i ← 1 TO n DO
    total ← total + nilai[i]
END FOR

rata_rata ← total / n

OUTPUT rata_rata

```python
# Menghitung rata-rata nilai ujian

n = int(input("Masukkan jumlah nilai ujian: "))

nilai = []
for i in range(n):
    angka = float(input(f"Masukkan nilai ke-{i + 1}: "))
    nilai.append(angka)

total = sum(nilai)
rata_rata = total / n

print("Daftar nilai:", nilai)
print("Total nilai:", total)
print(f"Rata-rata nilai ujian: {rata_rata:.2f}")
```

### Soal 2: Menentukan Bilangan Prima

Membuat algoritma untuk menentukan apakah suatu bilangan merupakan bilangan prima atau bukan.

INPUT n

IF n < 2 THEN
    OUTPUT "Bukan bilangan prima"
ELSE
    prima ← TRUE

    FOR i ← 2 TO n - 1 DO
        IF n MOD i = 0 THEN
            prima ← FALSE
            BREAK
        END IF
    END FOR

    IF prima = TRUE THEN
        OUTPUT "Bilangan prima"
    ELSE
        OUTPUT "Bukan bilangan prima"
    END IF
END IF

```python
# Menentukan apakah suatu bilangan merupakan bilangan prima

n = int(input("Masukkan bilangan: "))

if n < 2:
    prima = False
else:
    prima = True

    for i in range(2, n):
        if n % i == 0:
            prima = False
            break

if prima:
    print(n, "adalah bilangan prima")
else:
    print(n, "bukan bilangan prima")
```

### Soal 4: Trace Table Perhitungan Rata-rata

Diketahui nilai ujian: 70, 85, dan 90.

| Iterasi (i) | Nilai[i] | Total sebelum | Total sesudah |
|---|---:|---:|---:|
| 1 | 70 | 0 | 70 |
| 2 | 85 | 70 | 155 |
| 3 | 90 | 155 | 245 |

Perhitungan rata-rata:

Rata-rata = Total nilai / Jumlah nilai

Rata-rata = 245 / 3 = 81,67

Jadi, rata-rata nilai ujian adalah **81,67**

### Soal 5: Trace Algoritma Bilangan Prima

Diketahui n = 7.

| Iterasi | i | 7 MOD i | Keputusan |
|---|---:|---:|---|
| 1 | 2 | 1 | Tidak habis dibagi |
| 2 | 3 | 1 | Tidak habis dibagi |
| 3 | 4 | 3 | Tidak habis dibagi |
| 4 | 5 | 2 | Tidak habis dibagi |
| 5 | 6 | 1 | Tidak habis dibagi |

Karena tidak ada bilangan dari 2 sampai 6 yang membagi 7 secara habis, variabel prima tetap bernilai TRUE.

Kesimpulan: **7 adalah bilangan prima.**
