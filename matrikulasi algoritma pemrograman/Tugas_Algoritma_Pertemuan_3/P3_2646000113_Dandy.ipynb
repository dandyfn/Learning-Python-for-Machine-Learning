## Bagian A — Praktik Pengelola Nilai Interaktif


```python
# Nama: Dandy Al-Farisi Natanegara
# NIM: 646000113

def input_nilai():
    """Meminta nilai dan memastikan nilainya berada pada rentang 0-100."""
    try:
        teks = input("Masukkan nilai (0-100): ")

        if teks.strip() == "":
            raise ValueError("Input tidak boleh kosong.")

        nilai = float(teks)

        if nilai < 0 or nilai > 100:
            raise ValueError("Nilai harus berada antara 0 dan 100.")

    except ValueError as error:
        print(f"Input tidak valid: {error}")
        return None

    else:
        return nilai

    finally:
        print("Proses input selesai.")


def hitung_rata_rata(daftar_nilai):
    """Menghitung rata-rata menggunakan perulangan for."""
    total = 0

    for nilai in daftar_nilai:
        total += nilai

    return total / len(daftar_nilai)


def cari_nilai_tertinggi(daftar_nilai):
    """Mencari nilai tertinggi menggunakan perulangan for."""
    tertinggi = daftar_nilai[0]

    for nilai in daftar_nilai:
        if nilai > tertinggi:
            tertinggi = nilai

    return tertinggi


def main():
    daftar_nilai = []

    while True:
        print("\n=== PENGELOLA NILAI INTERAKTIF ===")
        print("1. Tambah nilai")
        print("2. Lihat rata-rata")
        print("3. Lihat nilai tertinggi")
        print("0. Keluar")

        pilihan = input("Pilih menu: ").strip()

        if pilihan == "1":
            nilai = input_nilai()

            if nilai is None:
                continue

            daftar_nilai.append(nilai)
            print(f"Nilai {nilai:g} berhasil ditambahkan.")

        elif pilihan == "2":
            if len(daftar_nilai) == 0:
                print("Belum ada nilai. Tambahkan nilai terlebih dahulu.")
                continue

            rata_rata = hitung_rata_rata(daftar_nilai)
            print(f"Rata-rata nilai: {rata_rata:.2f}")

        elif pilihan == "3":
            if len(daftar_nilai) == 0:
                print("Belum ada nilai. Tambahkan nilai terlebih dahulu.")
                continue

            tertinggi = cari_nilai_tertinggi(daftar_nilai)
            print(f"Nilai tertinggi: {tertinggi:g}")

        elif pilihan == "0":
            print("Program selesai. Terima kasih!")
            break

        else:
            print("Pilihan tidak valid. Masukkan 1, 2, 3, atau 0.")


if __name__ == "__main__":
    main()
```

### Nomor 2 — Pemilihan Nama Hari dengan match-case

```python
# Nomor 2: Pemilihan nama hari menggunakan match-case

hari = int(input("Masukkan nomor hari (1-7): "))

match hari:
    case 1:
        print("Senin")
    case 2:
        print("Selasa")
    case 3:
        print("Rabu")
    case 4:
        print("Kamis")
    case 5:
        print("Jumat")
    case 6:
        print("Sabtu")
    case 7:
        print("Minggu")
    case _:
        print("Nomor hari tidak valid.")
```


### Nomor 3 — Analisis Kompleksitas Waktu (Big-O)

Berdasarkan kerangka program Pengelola Nilai Interaktif pada PPT, terdapat perulangan `while True` yang digunakan untuk menampilkan menu secara berulang sampai pengguna memilih keluar. Kompleksitas waktu program bergantung pada jumlah pilihan menu yang dilakukan pengguna serta operasi yang dijalankan pada setiap pilihan.

**Analisis kode:**

1. **Perulangan menu (`while True`)**  
   Perulangan terus berjalan selama pengguna belum memilih keluar. Jika pengguna melakukan sebanyak k kali pemilihan menu, bagian ini berjalan sebanyak k iterasi. Dengan demikian, kompleksitasnya adalah O(k).

2. **Penambahan nilai (`nilai.append(...)`)**  
   Penambahan satu nilai ke dalam list umumnya memiliki kompleksitas waktu amortisasi O(1), yaitu konstan untuk setiap penambahan.

3. **Perhitungan rata-rata (`sum(nilai) / len(nilai)`)**  
   Fungsi `sum()` memeriksa seluruh elemen dalam list. Jika terdapat n nilai, kompleksitas perhitungannya adalah O(n). Namun, instruksi tugas meminta rata-rata dihitung menggunakan perulangan `for`, bukan `sum()`.

**Kesimpulan:** Kompleksitas waktu program secara keseluruhan tidak dapat ditentukan hanya dari satu potongan kode tanpa mengetahui jumlah pilihan menu dan operasi yang dijalankan. Untuk satu kali perhitungan rata-rata, kompleksitasnya adalah O(n), dengan n sebagai jumlah nilai yang tersimpan.


### Nomor 4 — Fungsi baca_umur()

```python

def baca_umur():
    while True:
        try:
            umur = int(input("Masukkan umur (0-120): "))

            if umur < 0 or umur > 120:
                raise ValueError("Umur harus antara 0 sampai 120.")

            return umur

        except ValueError as e:
            print("Input tidak valid:", e)


umur_pengguna = baca_umur()
print("Umur yang dimasukkan:", umur_pengguna)
```

### Nomor 5 — Exception Kustom NilaiTidakValid

```python

# Nomor 5: Membuat exception khusus

class NilaiTidakValid(Exception):
    pass


def tambah_nilai():
    try:
        nilai = float(input("Masukkan nilai (0-100): "))

        if nilai < 0 or nilai > 100:
            raise NilaiTidakValid("Nilai harus berada antara 0 sampai 100.")

        print("Nilai berhasil ditambahkan:", nilai)

    except ValueError:
        print("Input tidak valid. Masukkan angka.")

    except NilaiTidakValid as e:
        print("Kesalahan:", e)


tambah_nilai()
```
