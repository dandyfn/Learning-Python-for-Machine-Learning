## Bagian A – Praktik

```python
import time
import matplotlib.pyplot as plt

# Linear Search
def linear_search(data, target):
    for i in range(len(data)):
        if data[i] == target:
            return i
    return -1

# Binary Search untuk data terurut
def binary_search(data, target):
    kiri = 0
    kanan = len(data) - 1

    while kiri <= kanan:
        tengah = (kiri + kanan) // 2

        if data[tengah] == target:
            return tengah
        elif data[tengah] < target:
            kiri = tengah + 1
        else:
            kanan = tengah - 1

    return -1

# Ukuran data sesuai instruksi tugas
ukuran_data = [10**3, 10**4, 10**5, 10**6]
waktu_linear = []
waktu_binary = []
target = -1  # Target tidak ada dalam data

for n in ukuran_data:
    data = list(range(n))

    # Mengukur Linear Search
    mulai = time.perf_counter()
    hasil_linear = linear_search(data, target)
    durasi_linear = time.perf_counter() - mulai

    # Mengukur Binary Search
    mulai = time.perf_counter()
    hasil_binary = binary_search(data, target)
    durasi_binary = time.perf_counter() - mulai

    waktu_linear.append(durasi_linear)
    waktu_binary.append(durasi_binary)

    assert hasil_linear == -1
    assert hasil_binary == -1

# Menampilkan tabel hasil pengukuran
print(f"{'n':>10} | {'Linear Search (detik)':>23} | "
      f"{'Binary Search (detik)':>23}")
print("-" * 65)

for i in range(len(ukuran_data)):
    print(f"{ukuran_data[i]:>10} | "
          f"{waktu_linear[i]:>23.8f} | "
          f"{waktu_binary[i]:>23.8f}")

# Membuat dan menyimpan grafik
plt.figure(figsize=(9, 5))
plt.plot(ukuran_data, waktu_linear, marker="o",
         label="Linear Search")
plt.plot(ukuran_data, waktu_binary, marker="s",
         label="Binary Search")

plt.xscale("log")
plt.yscale("log")
plt.xlabel("Ukuran input (n)")
plt.ylabel("Waktu eksekusi (detik)")
plt.title("Perbandingan Waktu Linear Search dan Binary Search")
plt.legend()
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.tight_layout()

plt.savefig("P2_646000113_Dandy.png", dpi=300, bbox_inches="tight")
plt.show()
```
